```python
import numpy as np
import os
```

```python
output_base_dir = '/nexus/posix0/MBR-neuralsystems/ephys/GLC7189/GLC7189_c_g0'
outlier_dir = os.path.join(output_base_dir, 'outliers_list_sqr_overview.txt')
outlier_list = np.loadtxt(outlier_dir, delimiter=',')
print(outlier_list[:100])
```

```python
gap_threshold = 0.02 
min_duration = 0.01

gaps = np.diff(outlier_list)
split_indices = np.where(gaps > gap_threshold)[0]

eInkOn_raw = np.concatenate(([outlier_list[0]], outlier_list[split_indices + 1]))
eInkOff_raw = np.concatenate((outlier_list[split_indices], [outlier_list[-1]]))

durations = eInkOff_raw - eInkOn_raw
valid_mask = durations >= min_duration

eInkOn = eInkOn_raw[valid_mask]
eInkOff = eInkOff_raw[valid_mask]

print(f"Total points processed: {len(outlier_list)}")
print(f"Chunks found: {len(eInkOn)}")
print(f"Short chunks discarded: {np.sum(~valid_mask)}")

for i in range(len(eInkOn)):
    print(f"Chunk {i+1}: Start={eInkOn[i]:.8f}, End={eInkOff[i]:.8f}, Duration={eInkOff[i]-eInkOn[i]:.8f}s")

on_path = os.path.join(output_base_dir, 'eInkOn.txt')
off_path = os.path.join(output_base_dir, 'eInkOff.txt')

np.savetxt(on_path, eInkOn, fmt='%.8f')
np.savetxt(off_path, eInkOff, fmt='%.8f')

print(f"Successfully saved {len(eInkOn)} onsets to: {on_path}")
print(f"Successfully saved {len(eInkOff)} offsets to: {off_path}")
```
